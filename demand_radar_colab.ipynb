### 3) Run it
Swap `CONFIG` to target a different product, or write your own `configs/<product>.yaml`.


```python
CONFIG      = 'configs/ai_test_writer.yaml'   # real Hacker News + Stack Overflow data
# CONFIG    = 'configs/shortform_video.yaml'  # offline seed demo (no network)
LIVE_INGEST = True     # pull real posts from the configured sources
MAX_POSTS   = 20       # cap to respect the Gemini free-tier daily quota

from demand_radar.config import Config
from demand_radar import pipeline, report
from IPython.display import HTML, display

cfg = Config.load(CONFIG)
res = pipeline.run(cfg, live_ingest=LIVE_INGEST, max_posts=MAX_POSTS, outdir='outputs')
display(HTML(report.render_html(res.config, res.segments, res.leads, res.meta)))
```
